```python
import pandas as pd

df = pd.read_csv("ev_market_master.csv")
df.head()
```

```python
df.head()
df.info()
df.describe()
```

```python
# remove duplicates
df = df.drop_duplicates()

# check missing values
df.isnull().sum()
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df['units_sold'].sum()
```

```python
df.groupby('brand')['units_sold'].sum().sort_values(ascending=False)
```

```python
df.groupby('country')['units_sold'].sum().sort_values(ascending=False)
```

```python
df['year'] = df['date'].dt.year

df.groupby('year')['units_sold'].sum()
```

```python
import matplotlib.pyplot as plt

df.groupby('brand')['units_sold'].sum().plot(kind='bar')
plt.title("EV Sales by Brand")
plt.show()
```

```python
df[['units_sold','gasoline_price_usd_per_liter']].corr()
```

```python
df[['units_sold','total_chargers_cumulative']].corr()
```

```python
import matplotlib.pyplot as plt

brand_sales = df.groupby('brand')['units_sold'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
bars = plt.bar(brand_sales.index, brand_sales.values)

plt.title("EV Sales by Brand", fontsize=14, fontweight='bold')
plt.xlabel("Brand")
plt.ylabel("Units Sold")

# Add value labels
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval, int(yval),
             ha='center', va='bottom', fontsize=9)

plt.xticks(rotation=0)  # make labels straight
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(df['gasoline_price_usd_per_liter'], df['units_sold'])

plt.title("Fuel Price vs EV Sales", fontsize=14, fontweight='bold')
plt.xlabel("Fuel Price (USD per Liter)")
plt.ylabel("EV Units Sold")

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np

x = df['gasoline_price_usd_per_liter']
y = df['units_sold']

z = np.polyfit(x, y, 1)
p = np.poly1d(z)

plt.figure(figsize=(8,5))
plt.scatter(x, y)
plt.plot(x, p(x))

plt.title("Fuel Price vs EV Sales (Trend)", fontsize=14, fontweight='bold')
plt.xlabel("Fuel Price")
plt.ylabel("EV Units Sold")

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

brand_sales = df.groupby('brand')['units_sold'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))

bars = plt.bar(brand_sales.index, brand_sales.values)

# Title & labels
plt.title("EV Sales by Brand", fontsize=16, fontweight='bold')
plt.xlabel("Brand", fontsize=12)
plt.ylabel("Units Sold", fontsize=12)

# Remove top & right borders
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

# Add grid (light)
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Add value labels
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval,
             f'{yval/1e6:.1f}M',
             ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

brand_sales = df.groupby('brand')['units_sold'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))

bars = plt.bar(brand_sales.index, brand_sales.values)

# Make first bar highlight (important!)
bars[0].set_color('#2E86C1')   # highlight color
for bar in bars[1:]:
    bar.set_color('#A9CCE3')   # lighter color

# Clean style
plt.title("EV Sales by Brand", fontsize=16, fontweight='bold')
plt.xlabel("Brand")
plt.ylabel("Units Sold")

plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.grid(axis='y', linestyle='--', alpha=0.3)

# Value labels
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval,
             f'{yval/1e6:.1f}M',
             ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Prepare data
brand_sales = df.groupby('brand')['units_sold'].sum().sort_values(ascending=False)

# Create figure with 2 plots
fig, axes = plt.subplots(1, 2, figsize=(14,5))

# -----------------------
# 1. Bar Chart (Left)
# -----------------------
bars = axes[0].bar(brand_sales.index, brand_sales.values)

# Highlight top brand
bars[0].set_color('#2E86C1')
for bar in bars[1:]:
    bar.set_color('#A9CCE3')

axes[0].set_title("EV Sales by Brand", fontsize=14, fontweight='bold')
axes[0].set_xlabel("Brand")
axes[0].set_ylabel("Units Sold")

axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)
axes[0].grid(axis='y', linestyle='--', alpha=0.3)

# Value labels
for bar in bars:
    yval = bar.get_height()
    axes[0].text(bar.get_x() + bar.get_width()/2, yval,
                 f'{yval/1e6:.1f}M',
                 ha='center', va='bottom', fontsize=9)

# -----------------------
# 2. Scatter Plot (Right)
# -----------------------
x = df['gasoline_price_usd_per_liter']
y = df['units_sold']

axes[1].scatter(x, y, alpha=0.6)

# Trend line
z = np.polyfit(x, y, 1)
p = np.poly1d(z)
axes[1].plot(x, p(x))

axes[1].set_title("Fuel Price vs EV Sales", fontsize=14, fontweight='bold')
axes[1].set_xlabel("Fuel Price")
axes[1].set_ylabel("Units Sold")

axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)
axes[1].grid(alpha=0.3)

# -----------------------
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Prepare data
brand_sales = df.groupby('brand')['units_sold'].sum().sort_values(ascending=False)
yearly_sales = df.groupby('year')['units_sold'].sum()

# Create dashboard layout
fig, axes = plt.subplots(1, 3, figsize=(18,5))

# -----------------------
# 1. EV Sales by Brand
# -----------------------
bars = axes[0].bar(brand_sales.index, brand_sales.values)

# Highlight top brand
bars[0].set_color('#2E86C1')
for bar in bars[1:]:
    bar.set_color('#A9CCE3')

axes[0].set_title("EV Sales by Brand", fontweight='bold')
axes[0].set_xlabel("Brand")
axes[0].set_ylabel("Units Sold")
axes[0].grid(axis='y', alpha=0.3)

# Value labels
for bar in bars:
    yval = bar.get_height()
    axes[0].text(bar.get_x() + bar.get_width()/2, yval,
                 f'{yval/1e6:.1f}M',
                 ha='center', va='bottom', fontsize=9)

# -----------------------
# 2. Fuel Price vs EV Sales
# -----------------------
x = df['gasoline_price_usd_per_liter']
y = df['units_sold']

axes[1].scatter(x, y, alpha=0.6)

# Trend line
z = np.polyfit(x, y, 1)
p = np.poly1d(z)
axes[1].plot(x, p(x))

axes[1].set_title("Fuel Price vs EV Sales", fontweight='bold')
axes[1].set_xlabel("Fuel Price")
axes[1].set_ylabel("Units Sold")
axes[1].grid(alpha=0.3)

# -----------------------
# 3. Yearly Trend
# -----------------------
axes[2].plot(yearly_sales.index, yearly_sales.values, marker='o')

axes[2].set_title("EV Sales Trend", fontweight='bold')
axes[2].set_xlabel("Year")
axes[2].set_ylabel("Units Sold")
axes[2].grid(alpha=0.3)

# -----------------------
plt.tight_layout()
plt.show()
```
